## Day 3 Homework: Understanding and Handling Validation Errors

**Purpose:** Test invalid questions, check that each is rejected with the expected message, and then validate a corrected question. This exercise runs independently using Python's standard library; no model or package installation is needed.

### What the next cell does

1. **Defines `validate_public_question(question)`.** The function rejects non-string values, empty text, and whitespace-only text. It also rejects questions longer than 500 characters. Valid questions are returned with surrounding whitespace removed. The length check happens before trimming.
2. **Defines `run_error_examples()`.** This function prepares three deliberately invalid inputs and checks each one inside a `try` block.
3. **Checks the expected errors.** `raise ValueError` signals invalid input, and `except ValueError` catches it. An `assert` checks that the error message matches the expected message. If an invalid input is accepted, the test raises `AssertionError`.
4. **Validates a corrected question.** The example checks that a valid question with surrounding spaces becomes `How can I check my application status?`. This validates the question's format; it does not generate an answer.
5. **Runs the exercise and saves the report.** `result = run_error_examples()` starts the checks. The final lines create the `results` folder if needed and save the report to `results/errors_result.json`.

### Expected results

| Input | Reason for rejection | Expected message |
|---|---|---|
| `None` | The input is not a string | `Enter a non-empty question string.` |
| `'   '` | The text contains only whitespace | `Enter a non-empty question string.` |
| `'x' * 501` | The input exceeds 500 characters | `Use at most 500 characters.` |

The printed report should contain three entries under `intentional_caught_errors`, each with `expected_rejection: true`, followed by `corrected_input: "accepted"` and `checks: "passed"`. The last printed line should be `Saved results/errors_result.json`.

**The three caught errors indicate successful tests:** these inputs are deliberately invalid and should be rejected. An unhandled traceback or a failed assertion would instead indicate a problem that needs investigation.

Run the next code cell using ▶. To download the generated report in Colab, open the Files sidebar, expand `results`, and select Download for `errors_result.json`.


```python
# Adapted from Meaad Al-Marri’s SDA-AIE-213 extensions/error_examples.ipynb.
# Prediction: None and blank text are rejected; 501 characters exceeds the limit.
# raise signals invalid input; try runs validation; except ValueError catches the expected rejection.
# Corrected question: How can I check my application status?
# Deliberate, caught errors. These are examples to diagnose, not broken setup.
import json


def validate_public_question(question):
    if not isinstance(question, str) or not question.strip():
        raise ValueError('Enter a non-empty question string.')
    if len(question) > 500:
        raise ValueError('Use at most 500 characters.')
    return question.strip()


def run_error_examples():
    cases = [(None, 'Enter a non-empty question string.'),
             ('   ', 'Enter a non-empty question string.'),
             ('x' * 501, 'Use at most 500 characters.')]
    rows = []
    for value, expected in cases:
        try:
            validate_public_question(value)
        except ValueError as error:
            assert str(error) == expected
            rows.append({'error_type': type(error).__name__, 'message': str(error), 'expected_rejection': True})
        else:
            raise AssertionError('Invalid input was accepted.')
    assert validate_public_question(' How can I check my application status? ') == 'How can I check my application status?'
    result = {'intentional_caught_errors': rows, 'corrected_input': 'accepted', 'checks': 'passed'}
    print(json.dumps(result, indent=2))
    return result



result = run_error_examples()

from pathlib import Path
output_dir = Path('results')
output_dir.mkdir(exist_ok=True)
(output_dir / 'errors_result.json').write_text(json.dumps(result, indent=2), encoding='utf-8')
print('Saved results/errors_result.json')
```
